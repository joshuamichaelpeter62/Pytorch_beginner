## Rules that perform matrix multiplication
1. The inner dimensions match
* '(3, 2) @ (3, 2)' won't work
* '(2, 3) @ (3, 2)' will work
* '(3, 2) @ (2, 3)' will work
2. The resulting matrix has the shape of the **outer dimensions**:
* '(2, 3) @ (3, 2)' -> "(2, 2)'
* '(3, 2) @ (2, 3)' -> '(3, 3)'
