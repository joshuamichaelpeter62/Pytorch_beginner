## Rules that perform matrix multiplication
1. The inner dimensions match
* '(3, 2) @ (3, 2)' won't work
* '(2, 3) @ (3, 2)' will work
* '(3, 2) @ (2, 3)' will work
2. The resulting matrix has the shape of the **outer dimensions**:
* '(2, 3) @ (3, 2)' -> "(2, 2)'
* '(3, 2) @ (2, 3)' -> '(3, 3)'

```python
torch.matmul(torch.rand(3,10), torch.rand(10,3))
```

# reshaping, stacking, squeezing and unsqueezing tensor
* Reshaping: reshaping an input tensor to a defend shape
* View: Return a view of an input tensor of a certain shape but keep the same memory as the original tensor
* Stacking: combine multiple tensor on top of each other (vstack) or side (hstack)
* Squeeze: Remove dimension from tensor
* unsqueeze: add dimension to a target tensor
* Permute: Return a view of the input with dimensions permuted

```python
# squeeze and unsqueeze
squ = torch.zeros(2, 1, 2, 1, 2)
squ.size()
```

```python
y_squ = torch.squeeze(squ)
y_squ.size()
```

```python
y_squ = torch.squeeze(squ, 0)
y_squ.size()
```

```python
y_squ = torch.squeeze(squ, 1)
y_squ.size()
```

```python
y_squ = torch.squeeze(squ, (1,2,3))
y_squ.size()
```

```python
# torch.unsqueeze() - add single dimesion to a target tensor at specific dim
print(f"previous target: {y_squ.shape}")
y_unsqu = y_squ.unsqueeze(dim=0)
print(f"new target: {y_unsqu.shape}")
```

```python
# torch.permute - rearrange the dimesion of a target tensor in specific order
x_original = torch.rand(size=(224,224,3))
x_original

# permut the orginal tensor to rearrange the axis (or dim) order
x_permuted = x_original.permute(2, 0, 1) # shifts axis 0->1, 1->2, 2->0
print(f"previous shape: {x_original.shape}")
print(f"new shape: {x_permuted.shape}") # color channel height and width
```
